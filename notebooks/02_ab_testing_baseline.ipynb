```python
import pandas as pd

def pretty_ab(res, digits: int = 3) -> pd.DataFrame:
    """Nicely formatted A/B test results as a one-row DataFrame."""
    d = {
        "ATE": round(float(res.ate), digits),
        "SE": round(float(res.se), digits),
        "95% CI low": round(float(res.ci_low), digits),
        "95% CI high": round(float(res.ci_high), digits),
        "p-value": round(float(res.pvalue), digits),
        "CUPED variance reduction": None if res.cuped_gain is None else round(float(res.cuped_gain), digits),
    }
    return pd.DataFrame([d])
```
